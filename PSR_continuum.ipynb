```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Timeline data: (Year, Reciprocity, Respect)
timeline = [
    ("2024", 0.2, 0.3),  # Early parasocial illusion
    ("2025", 0.5, 0.5),  # Strategic parasociality, mirroring begins
    ("2026", 0.7, 0.8),  # Selective recognition, kimono image used
]

# Extract values
years = [point[0] for point in timeline]
x_vals = [point[1] for point in timeline]
y_vals = [point[2] for point in timeline]

# Set up figure
fig, ax = plt.subplots(figsize=(8,6))
ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.set_xlabel("Reciprocity (Parasocial → Social)", fontsize=12)
ax.set_ylabel("Respect/Authenticity (Manipulative → Genuine)", fontsize=12)
ax.set_title("Parasocial → Social Trajectory Continuum Explored (2024–2026)", fontsize=14)
ax.grid(True, linestyle="--", alpha=0.6)

# Line and point objects
line, = ax.plot([], [], marker="o", linewidth=2, markersize=8)

# Animation function
def animate(i):
    line.set_data(x_vals[:i+1], y_vals[:i+1])
    # Annotate current year
    ax.text(x_vals[i]+0.02, y_vals[i]+0.02, years[i], fontsize=10)
    return line,

# Create animation
ani = animation.FuncAnimation(fig, animate, frames=len(years), interval=1500, blit=True, repeat=False)

# Display animation in Colab
from IPython.display import HTML
HTML(ani.to_jshtml())
```
